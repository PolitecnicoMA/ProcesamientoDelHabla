```python
import requests
from bs4 import BeautifulSoup

def scrapear_web(url):
    response = requests.get(url)
    soup = BeautifulSoup(response.text, "html.parser")

    parrafos = soup.find_all("p")

    texto = ""
    for p in parrafos:
        texto += p.get_text()

    return texto
```

```python
def dividir_texto(texto, tamaño=100):
    palabras = texto.split()
    return [" ".join(palabras[i:i+tamaño]) for i in range(0, len(palabras), tamaño)]
```

```python
fragmentos = dividir_texto(corpus)
```

```python
!pip install sentence-transformers
```

```python
from sentence_transformers import SentenceTransformer

#modelo = SentenceTransformer('all-MiniLM-L6-v2')
modelo = SentenceTransformer('paraphrase-multilingual-MiniLM-L12-v2')

vectores = modelo.encode(fragmentos)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

def buscar(query, top_k=3):
    query_emb = modelo.encode([query])  #

    similitudes = cosine_similarity(query_emb, vectores)[0]

    top_indices = np.argsort(similitudes)[-top_k:][::-1]

    for i in top_indices:
        print("\n Resultado:")
        print(fragmentos[i][:300])
```
