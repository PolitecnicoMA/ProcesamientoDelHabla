```python
import wikipedia

wikipedia.set_lang("es")
texto_wiki = wikipedia.page("Minecraft").content
```

```python
import PyPDF2

def leer_pdf(path):
    texto = ""
    with open(path, "rb") as file:
        reader = PyPDF2.PdfReader(file)
        for page in reader.pages:
            texto += page.extract_text()
    return texto

texto_libro = leer_pdf("100as.pdf")
```

```python
corpus = texto_wiki + "\n" + texto_libro
```

```python
def dividir_texto(texto, tamaño=200):
    palabras = texto.split()
    return [" ".join(palabras[i:i+tamaño]) for i in range(0, len(palabras), tamaño)]
```

```python
fragmentos = dividir_texto(corpus)
```

```python
!pip install sentence-transformers
```

```python
from sentence_transformers import SentenceTransformer

modelo = SentenceTransformer('all-MiniLM-L6-v2')

vectores = modelo.encode(fragmentos)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

def buscar(query, top_k=3):
    query_emb = modelo.encode([query])  #

    similitudes = cosine_similarity(query_emb, vectores)[0]

    top_indices = np.argsort(similitudes)[-top_k:][::-1]

    for i in top_indices:
        print("\n Resultado:")
        print(fragmentos[i][:100])
```

```python
buscar("un juego donde podes construir ")
```

```python
buscar("familia buendia ")
```

```python
buscar("tecnologia y videojuegos")
```
